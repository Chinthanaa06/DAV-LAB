```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\DELL\chinthana\archive (1)\student_dropout_risk_2026.csv")
df.head()
```

```python
df.columns
```

```python
plt.hist(df['satisfaction_score'], bins=10, edgecolor='black')
plt.xlabel('Satisfaction Score')
plt.ylabel('Frequency')
plt.title('Histogram of Satisfaction Score')
plt.show()
```

```python
plt.boxplot(df['satisfaction_score'].dropna())
plt.ylabel('Satisfaction Score')
plt.title('Box Plot of Satisfaction Score')
plt.show()
```

```python
plt.scatter(df['attendance_rate'], df['satisfaction_score'])
plt.xlabel('Attendance Rate')
plt.ylabel('Satisfaction Score')
plt.title('Attendance Rate vs Satisfaction Score')
plt.show()
```
