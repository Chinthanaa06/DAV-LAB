```python
import numpy as np
arr=np.array([1,2,3,4,5])
print("Numpy Array")
print(arr)
print(arr[3])
print(arr[-3])
print(arr[2:3])
print(arr[-3:])
print(arr+3)
print(arr-3)
print(arr*3)
print(arr/2)
print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(arr.reshape(1,5))
```
